```python
import numpy as np
import os
import cv2 
```

```python
import ipywidgets as widgets
from IPython.display import display
from PIL import Image
import matplotlib.pyplot as plt

def show_bmp(n):
    
    plt.figure(figsize=(10, 10))



    plt.subplot(1, 2, 1)
    filename = os.path.join("output", "input", f"{n}.bmp")
    img = np.array(Image.open(filename))
    plt.imshow(img[1::2])
    plt.axis("off")
    plt.title(f"Displaying: {filename}")

    plt.subplot(1, 2, 2)
    filename = os.path.join("output", "input", f"{n}.bmp")
    img = np.array(Image.open(filename))
    plt.imshow(img[1::2])
    plt.axis("off")
    plt.title(f"Displaying: {filename}")


    plt.show()
    
    
# Create an interactive slider for file number (adjust min and max as needed)
slider = widgets.IntSlider(value=0, min=0, max=300, step=1, description="File Number:")

# Connect the slider to the show_bmp function
interactive_widget = widgets.interactive(show_bmp, n=slider)
display(interactive_widget)
```
