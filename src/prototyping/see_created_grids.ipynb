```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
dir = os.path.join("..", "neural_network", "data", "upsampled_train")
paths = [os.path.join(dir, p) for p in os.listdir(dir)]

paths
```

```python
import imageio

with imageio.get_writer("events_render.mp4", fps=1) as f:
    for frame in grids:
        for g in frame:
            f.append_data(g)

```
