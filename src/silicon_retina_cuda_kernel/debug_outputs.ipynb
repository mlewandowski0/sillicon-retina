```python
def get_from_folder(folder):
    els = sorted([(int(f[:-4]), os.path.join("out", folder, f)) for f in os.listdir(os.path.join("out", folder))])
    return [np.load(v) for _, v in els]

imgs = get_from_folder("imgs")
lpfs1 = get_from_folder("LPF1")
lpfs2 = get_from_folder("LPF2")
hpf1 = get_from_folder("HPF1")
hpf2 = get_from_folder("HPF2")
blur2 = get_from_folder("BLUR2")
I_opl = get_from_folder("I_OPL")
Vlif = get_from_folder("VLIF")
fired = get_from_folder("fired")
```
